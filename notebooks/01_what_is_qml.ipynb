Through the material, I have become somewhat familiar with Dirac notation representing a qubit's (or circuit of qubits') **state vector**.

In Dirac notation, a single qubit can be represented as:

$\psi = \alpha|0\rangle + \beta|1\rangle$ 

or 

$\psi = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}$

Where $\psi$ is the state vector,

$\alpha$ is the amplitude that determines the proability of the qubit measuring to 0,

and $\beta$ is the amplitude that determines the probability of the qubit measuring to 1.

This formula highlights the immutable existence of 2 distinct possiblities for any single qubit in any moment. 

A probability amplitude can be converted directly into percentages via the absolute value of amplitude squared or: 

$P(n)=|\langle n|\psi \rangle|^2$

Where $P(n)$ represents the real probability of a specific outcome n.

And $\langle n|\psi \rangle$ represents the probability amplitude.

I've heard that once you build a circuit that includes more than one qubit, you can no longer measure the value of just one. Which makes sense considering they are all "entangled" and the value of one becoming 1 or 0 would directly affect the value of all of them. Therefore, the measurements we take become the measure of the whole circuit. The whole circuit can be represented similarly to the single qubit notation, but with the added cross-qubit possibilities represented.
