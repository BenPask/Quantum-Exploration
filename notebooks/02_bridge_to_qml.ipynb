Tensor products are used to join the Hilbert space of multiple qubits together. It seems like they are neccessary for any visual representation of multiple qubits existing in the same space.

One thing that has been rather confusing for me is the Venn diagram of Tensor Products, Superposition, and Entanglement. The three are distinct concpets in quantum mechanics, but the lines between has slowly started to merge while I learned.

The easiest way to differentiate these concepts is that tensor products and superposition keep the qubits independent, but entanglement does not.

For example, given the state vector $\psi_1$ of a qubit:

$\psi_1 = \alpha_1|0\rangle + \beta_1|1\rangle$

and the state vector $\psi_2$ of another qubit:

$\psi_2 = \alpha_2|0\rangle + \beta_2|1\rangle$

We have not taken a tensor product, put them into superposition, or entagled them.

To get the tensor product of the two states:

$\psi_1 \otimes \psi_2 = \alpha_1\alpha_2|00\rangle + \alpha_1\beta_2|01\rangle + \beta_1\alpha_2|10\rangle + \beta_1\beta_2|11\rangle$

We have now combined the state vector of qubit 1 and qubit 2 via a tensor product. Importantly, though, we have NOT neccessarily put either into superposition (neither has a definitive state yet) and we definitely have NOT entangled the two qubits. If they were entangled, you could no longer seperate the values of the 2 qubits. Because we can clearly see where each qubits' values begin and end (labeled via $_1$ and $_2$), they are very clearly still independent of eachother.

Yet, this example still visualizes the 2^n nature of mapping multiple qubits, the true power of quantum computation.


## 2. Superposition

Using the same example, we can put the qubits into superposition by applying a Hadamard gate. Applying this gate is applying a 90 degree rotation to the qubits representation on the bloch sphere. However, before we go to apply that H-gate, we need to initialize the proability amplitudes of each qubit's state. To do this, we can simply initialize them into the standard North-Pole rotation that they start as when existent in near 0k temperatures:

$\psi_1 = 1.0|0\rangle + 0.0|1\rangle$

$\psi_2 = 1.0|0\rangle + 0.0|1\rangle$

These state vectors reflect a 100% proability of each independent qubit measuring 0 when measured. The tensor product of the state vectors becomes:

$\psi_1 \otimes \psi_2 = 1.0|00\rangle + 0.0|01\rangle + 0.0|10\rangle + 0.0|11\rangle$



These qubits now have definitive probability amplitudes, but because they do not hold multiple possibilities at once(both have a 100% chance of measuring 0), they are not yet in superposition. This is where the H-gate comes in. We'll start with the alpha and beta probability amplitude representation to understand the whole process better:

$H|\psi_1\rangle = \alpha_1(H|0\rangle) + \beta_1(H|1\rangle)$

$H|\psi_2\rangle = \alpha_2(H|0\rangle) + \beta_2(H|1\rangle)$

filling in the value of the Hadamard gates:

$H|\psi_1\rangle = \alpha_1(\frac{|0\rangle + |1\rangle}{\sqrt2}) + \beta_1(\frac{|0\rangle - |1\rangle}{\sqrt2})$

$H|\psi_2\rangle = \alpha_2(\frac{|0\rangle + |1\rangle}{\sqrt2}) + \beta_2(\frac{|0\rangle - |1\rangle}{\sqrt2})$

Contintuing multiplicative process:

$H|\psi_1\rangle = (\frac{\alpha_1 + \beta_1}{\sqrt2})|0\rangle + (\frac{\alpha_1 - \beta_1}{\sqrt2})|1\rangle$

$H|\psi_2\rangle = (\frac{\alpha_2 + \beta_2}{\sqrt2})|0\rangle + (\frac{\alpha_2 - \beta_2}{\sqrt2})|1\rangle$


Filling in our values for the probability amplitudes $\alpha$ and $\beta$:

$H|\psi_1\rangle = (\frac{1.0 + 0.0}{\sqrt2})|0\rangle + (\frac{1.0 - 0.0}{\sqrt2})|1\rangle$

$H|\psi_2\rangle = (\frac{1.0 + 0.0}{\sqrt2})|0\rangle + (\frac{1.0 - 0.0}{\sqrt2})|1\rangle$

Or:

$H|\psi_1\rangle = \frac{1}{\sqrt2}|0\rangle + \frac{1}{\sqrt2}|1\rangle$

$H|\psi_2\rangle = \frac{1}{\sqrt2}|0\rangle + \frac{1}{\sqrt2}|1\rangle$

Is the state vector of qubit 1 and qubit 2 after applying a Hamand gate. From this, we can see that it has become equally likely to measure 0 or 1 in measuring either qubit a single time. This gate has shifted each state vector's Bloch vector from straight up (North-Pole position) to pointing exactly at the equator of the Bloch sphere. This makes each qubit's state vector hold multiple possible measurements simultaneously, holding **superposition.**

To see how this affects our tensor product of the two vectors, we'll just plug in the new probability amplitudes into our existing tensor product to get:

$H|\psi_1\rangle \otimes H|\psi_2\rangle = (\frac{1}{\sqrt2})(\frac{1}{\sqrt2})|00\rangle + (\frac{1}{\sqrt2})(\frac{1}{\sqrt2})|01\rangle + (\frac{1}{\sqrt2})(\frac{1}{\sqrt2})|10\rangle + (\frac{1}{\sqrt2})(\frac{1}{\sqrt2})|11\rangle$

Continuing via multiplication:

$H|\psi_1\rangle \otimes H|\psi_2\rangle = \frac{1}{2}|00\rangle + \frac{1}{2}|01\rangle + \frac{1}{2}|10\rangle + \frac{1}{2}|11\rangle$

We now have the **tensor product** of 2 qubits' state vectors put into **superposition**, but they are still NOT entangled.

The CNOT gate is apparently a 2x2 matrix that is multiplied by a qubits state vector and entangles it with another qubit's state vector. Let's get a look at this matrix:

_Note: in searching for the standard CNOT matrix, I have realized that applying a CNOT gate only entangles 2 qubits if the control qubit is in a superposition state._

If the control qubit is not in a superposition state such as:

$\psi_c = 0.0|0\rangle + 1.0|1\rangle$ or $\psi_c = 1.0|0\rangle + 0.0|1\rangle$

$\psi_t = \alpha|0\rangle + \beta|1\rangle$

Then applying a CNOT gate (applying an X-gate only if the control qubit is 1) simply acts as a switch:

$\psi_t = \beta|0\rangle + \alpha|1\rangle$ or, remaining the same: $\alpha|0\rangle + \beta|1\rangle$ respectively

This does not cause entanglement, as no Bell state or dependence is created between anything.


The same applies to applying CNOT gates to qubits where BOTH qubits are in full superposition, because they still equal the exact same thing once the CNOT gate is applied.

$\psi_1 \otimes \psi_2 = \frac{1}{2}|00\rangle + \frac{1}{2}|01\rangle + \frac{1}{2}|10\rangle + \frac{1}{2}|11\rangle$

Applying CNOT, control = qubit 1 (apply X-gate to qubit 2 only if the control bit is 1): </br>
In our state vector, the control bit is currently independent and represented by the left digit of each possibility.

$\psi_1 \otimes \psi_2 = \frac{1}{2}|00\rangle + \frac{1}{2}|01\rangle + \frac{1}{2}|11\rangle + \frac{1}{2}|10\rangle$


Now, to create entanglement, we can simply apply initialize 2 qubits and only apply an H-gate to one of them. This puts one qubit in superposition and leaves the other in North-Pole:

$\psi_1 = 1.0|0\rangle + 0.0|1\rangle$

$\psi_2 = 1.0|0\rangle + 0.0|1\rangle$

Apply H-gate to only qubit 1:

$\psi_1 = \frac{1}{\sqrt2}|0\rangle + \frac{1}{\sqrt2}|1\rangle$

$\psi_2 = 1.0|0\rangle + 0.0|1\rangle$

Get the tensor product of the 2 qubits' state vectors:

$\psi_1 \otimes \psi_2 = \frac{1}{\sqrt2}|00\rangle + 0.0|01\rangle + \frac{1}{\sqrt2}|10\rangle + 0.0|11\rangle$

Apply a CNOT gate on $\psi_1$ and $\psi_2$, with the controll being $\psi_1$ (apply X-gate to qubit 2 only if the control bit is 1):

$\psi_1 \otimes \psi_2 = \frac{1}{\sqrt2}|00\rangle + 0.0|01\rangle + \frac{1}{\sqrt2}|11\rangle + 0.0|10\rangle$

Put back in initial order:

$\Psi = \frac{1}{\sqrt2}|00\rangle + 0.0|01\rangle  + 0.0|10\rangle + \frac{1}{\sqrt2}|11\rangle$

We can see that the state vectors of our two qubits have become **entangled**, with the orignal representations of each qubit being lost in the combination (because the right bit is now reliant on the left, there is no seperation or individual representations of either qubit).

This is now a representation of a quantum circuit's vector space, and these exact values make up the **Bell State:**

**Bell State (matrix form)** = $\begin{pmatrix} \frac{1}{\sqrt2} \\ 0 \\ 0 \\ \frac{1}{\sqrt2} \end{pmatrix}$
